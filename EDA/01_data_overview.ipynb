
# Phase I Part 1 — Data Overview
This notebook loads the Airbnb NYC 2019 dataset and reports the structural diagnostics requested for Phase I.


```python

from pathlib import Path
import pandas as pd
from IPython.display import display

DATA_PATH = Path('AB_NYC_2019.csv')
df = pd.read_csv(DATA_PATH)
print(f"Loaded {DATA_PATH.resolve()}")
```

```python

print('Shape (rows, columns):', df.shape)
display(df.head(5))
```

**Note:** `df.shape` confirms the dataset has 48,895 rows and 16 columns; reviewing `df.head()` alongside it helps verify that column meanings and value ranges match the Airbnb NYC 2019 documentation.


```python

df.info()
```

**Note:** `df.info()` shows most columns are numeric or object types, while fields like `last_review` load as strings and must be converted before any time-based analysis.


```python

missing = df.isna().sum().to_frame(name='missing_count')
missing['missing_pct'] = (missing['missing_count'] / len(df) * 100).round(2)
display(missing)
```

**Note:** Missing values cluster in `reviews_per_month` and `last_review` (about 10k rows) with minor gaps in `host_name` and `name`, which motivates the minimal filling/dropping strategy in the cleaning notebook.
