
# Prediction Problem Definition
- **Outcome (y):** nightly listing `price` (continuous, USD) from the 2019 NYC Airbnb dataset.
- **Candidate numeric predictors:** `latitude`, `longitude`, `minimum_nights`, `number_of_reviews`, `reviews_per_month`, `calculated_host_listings_count`, `availability_365`.
- **Candidate categorical predictors:** `neighbourhood_group`, `room_type`.
- **Why it matters:** Accurate price expectations help hosts benchmark rates, guide guests toward fair listings, and let the platform surface competitive options without training full ML models at this phase.


**Note:** This notebook only documents the prediction target, candidate features, and pricing motivation so the rest of the workflow can reference a consistent problem statement.
