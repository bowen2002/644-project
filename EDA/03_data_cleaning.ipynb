
# Light Data Cleaning (Phase I)
Minimal transformations applied so later statistics and plots use consistent assumptions.


```python

from pathlib import Path
import numpy as np
import pandas as pd
from IPython.display import display

DATA_PATH = Path('AB_NYC_2019.csv')
df = pd.read_csv(DATA_PATH)
raw_rows = len(df)
```

```python

# Ensure price is numeric and handle missing reviews_per_month
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)

# Drop rows missing price
df = df.dropna(subset=['price']).copy()

# Log1p target for visualization purposes only
df['y_log'] = np.log1p(df['price'])

# Cap price at 99th percentile for plots (visual clarity only)
price_cap = df['price'].quantile(0.99)
df['price_for_viz'] = df['price'].clip(upper=price_cap)

cleaning_summary = {
    'rows_before_cleaning': raw_rows,
    'rows_after_cleaning': len(df),
    'price_missing_after': int(df['price'].isna().sum()),
    'reviews_per_month_missing_after': int(df['reviews_per_month'].isna().sum()),
    'price_cap_for_viz': float(price_cap)
}
cleaning_summary
```


Note: `price_for_viz` should be used only when plotting so extreme outliers (>$800) do not dominate axes. The original `price` column remains unchanged for any downstream modeling once Phase II begins.
