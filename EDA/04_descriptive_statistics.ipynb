
# Descriptive Statistics
Summary tables for price overall and by key categorical predictors.


```python

from pathlib import Path
import numpy as np
import pandas as pd
from IPython.display import display

DATA_PATH = Path('AB_NYC_2019.csv')

def load_clean(path=DATA_PATH):
    df = pd.read_csv(path)
    df['price'] = pd.to_numeric(df['price'], errors='coerce')
    df['reviews_per_month'] = df['reviews_per_month'].fillna(0)
    df = df.dropna(subset=['price']).copy()
    df['y_log'] = np.log1p(df['price'])
    price_cap = df['price'].quantile(0.99)
    df['price_for_viz'] = df['price'].clip(upper=price_cap)
    return df

df = load_clean()
print(f"Cleaned dataset rows: {len(df)}")
```

```python

price_summary = df['price'].describe()
display(price_summary.to_frame(name='price'))
```

```python

by_room = df.groupby('room_type')['price'].agg(['mean', 'median', 'count']).sort_values('mean', ascending=False)
by_neighbourhood = df.groupby('neighbourhood_group')['price'].agg(['mean', 'median', 'count']).sort_values('mean', ascending=False)

display(by_room)
display(by_neighbourhood)
```

**Note:** The grouped tables clearly separate mean/median prices by room type and borough—entire homes and Manhattan sit highest, while shared rooms and the Bronx sit lowest, so you can quote these differences directly.
